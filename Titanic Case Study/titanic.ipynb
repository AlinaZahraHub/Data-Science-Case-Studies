Task 15: Calculate Missing-Value Percentage

```python
missing_percentage = (df.isnull().sum() / len(df) * 100).round(2)
missing_df = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": missing_percentage
})
print(missing_df)
```

```python
# Which column has the highest missing percentage?: Cabin (approximately 77.10%).

# Is the missing percentage in Age small or significant?: It is moderate/significant (~19.87% missing), meaning it needs careful handling.

# Does Embarked contain many missing values?: No, it contains only 2 missing values (~0.22%), which is very negligible.

# Would filling Cabin be reliable if most values are missing? Explain briefly.: No, filling Cabin would be entirely unreliable and arbitrary because over 77% of the data is missing. Fabricating or guessing that many records would introduce massive noise and distort real patterns.
```

Task 16: Remove Duplicate Rows

```python
df = df.drop_duplicates().copy()
```

Task 17: Handle Missing Age Values

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
```

Task 18: Handle Missing Embarked Values

```python
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
```

Task 19: Verify Missing Values After Cleaning

```python
print(df.isnull().sum())
```

```python
# Does Age still contain missing values?: No (0 missing values remaining).

# Does Embarked still contain missing values?: No (0 missing values remaining).

# Which column still contains a large number of missing values?: Cabin still contains a large number of missing values because we intentionally left it unchanged as per instructions.
```

Task 24: Detect Possible Outliers Using Boxplots

```python
numerical_cols = ["Age", "Fare", "SibSp", "Parch"]
plt.figure(figsize=(8, 4))
sns.boxplot(data=df[numerical_cols])
plt.title("Boxplots of Numerical Features")
plt.show()
```

```python
# Which feature has the most obvious outliers?: Fare has the most prominent extreme outliers.

# Does Fare contain unusually high observations?: Yes, several data points stretch far above the upper quartile, reaching up to 512.

# Does seeing an outlier automatically mean the value is incorrect? Explain briefly.: No, an outlier does not automatically mean a data error. In this dataset, high fares represent actual luxury suites purchased by wealthy first-class passengers.
```

Task 25: Create a Correlation Matrix

```python
selected_cols = ["Age", "SibSp", "Parch", "Pclass", "Fare"]

# Correlation Matrix
print(df[selected_cols].corr())

```

Task 26: Display Correlation as a Heatmap

```python
# Heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(df[selected_cols].corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```

```python
# Which two features show the strongest positive correlation?: SibSp and Parch show a notable positive correlation (families travelling together tend to have both spouses/siblings and parents/children).

# Which variables show a negative correlation?: Pclass and Fare show a strong negative correlation (since Pclass 1 is represented by lower numeric value 1 and high cost, while Pclass 3 is 3 with low cost).

# What relationship do you observe between Pclass and Fare?: Lower class numbers (1st class) correspond to significantly higher fares, whereas 3rd class corresponds to lower fares.

# Does correlation prove that one variable causes another?: No, correlation only measures statistical association, not direct causation.
```
